```python
# Load Gold Fact Tables

fact_vendor_performance =  spark.read.table("fact_vendor_performance")



fact_inventory = spark.read.table("fact_inventory")

fact_procurement = spark.read.table("fact_procurement")

print("Tables Loaded Successfully")
```

```python
print(fact_vendor_performance.columns)
print(fact_procurement.columns)
print(fact_inventory.columns)
```

```python
from pyspark.sql.functions import desc

display(
    fact_procurement
    .groupBy("VendorName")
    .sum("TotalPurchaseSpend")
    .orderBy(desc("sum(TotalPurchaseSpend)"))
    .limit(10)
)
```

```python
from pyspark.sql.functions import desc

display(
    fact_vendor_performance
    .groupBy("VendorName")
    .sum("GrossProfit")
    .orderBy(desc("sum(GrossProfit)"))
    .limit(10)
)
```

```python
from pyspark.sql.functions import avg

display(
    fact_vendor_performance
    .groupBy("VendorName")
    .agg(avg("ProfitMargin").alias("AvgProfitMargin"))
    .orderBy("AvgProfitMargin")
    .limit(10)
)
```

```python
from pyspark.sql.functions import avg, desc

display(
    fact_vendor_performance
    .groupBy("VendorName")
    .agg(avg("StockTurnover").alias("AvgStockTurnover"))
    .orderBy(desc("AvgStockTurnover"))
    .limit(10)
)
```

```python
from pyspark.sql.functions import desc

display(
    fact_inventory
    .groupBy("Description")
    .sum("InventoryValue")
    .orderBy(desc("sum(InventoryValue)"))
    .limit(10)
)
```

```python
from pyspark.sql.functions import desc

display(
    fact_inventory
    .groupBy("Store")
    .sum("InventoryValue")
    .orderBy(desc("sum(InventoryValue)"))
    .limit(10)
)
```

```python
from pyspark.sql.functions import desc

display(
    fact_procurement
    .groupBy("VendorName")
    .sum("TotalPurchaseQuantity")
    .orderBy(desc("sum(TotalPurchaseQuantity)"))
    .limit(10)
)
```

```python
from pyspark.sql.functions import desc

display(
    fact_procurement
    .groupBy("VendorName")
    .sum("TotalFreight")
    .orderBy(desc("sum(TotalFreight)"))
    .limit(10)
)
```

```python
from pyspark.sql.functions import sum

total_spend = fact_procurement.agg(
    sum("TotalPurchaseSpend").alias("TotalSpend")
).collect()[0]["TotalSpend"]

top3_spend = fact_procurement.groupBy("VendorName") \
    .sum("TotalPurchaseSpend") \
    .orderBy("sum(TotalPurchaseSpend)", ascending=False) \
    .limit(3) \
    .agg(sum("sum(TotalPurchaseSpend)").alias("Top3Spend")) \
    .collect()[0]["Top3Spend"]

percentage = (top3_spend / total_spend) * 100

print(f"Top 3 Vendors Spend % = {percentage:.2f}%")
```

```python
from pyspark.sql.functions import desc

top_spend = fact_procurement \
    .groupBy("VendorName") \
    .sum("TotalPurchaseSpend") \
    .orderBy(desc("sum(TotalPurchaseSpend)")) \
    .limit(10)

top_profit = fact_vendor_performance \
    .groupBy("VendorName") \
    .sum("GrossProfit") \
    .orderBy(desc("sum(GrossProfit)")) \
    .limit(10)

display(top_spend)
display(top_profit)
```

```python
from pyspark.sql.functions import desc

top_spend = fact_procurement \
    .groupBy("VendorName") \
    .sum("TotalPurchaseSpend") \
    .orderBy(desc("sum(TotalPurchaseSpend)")) \
    .limit(10)

top_freight = fact_procurement \
    .groupBy("VendorName") \
    .sum("TotalFreight") \
    .orderBy(desc("sum(TotalFreight)")) \
    .limit(10)

display(top_spend)
display(top_freight)
```

```python
from pyspark.sql.functions import col

display(
    fact_procurement
    .withColumn(
        "FreightPct",
        (col("TotalFreight") / col("TotalPurchaseSpend")) * 100
    )
    .groupBy("VendorName")
    .avg("FreightPct")
    .orderBy("avg(FreightPct)", ascending=False)
)
```

```python
from pyspark.sql.functions import avg, desc

display(
    fact_vendor_performance
    .groupBy("VendorName")
    .agg(avg("ProfitMargin").alias("AvgProfitMargin"))
    .orderBy(desc("AvgProfitMargin"))
    .limit(10)
)
```

```python
from pyspark.sql.functions import sum

spend = fact_procurement.groupBy("VendorName") \
    .agg(sum("TotalPurchaseSpend").alias("Spend"))

profit = fact_vendor_performance.groupBy("VendorName") \
    .agg(sum("GrossProfit").alias("Profit"))

display(
    spend.join(profit, "VendorName")
         .orderBy("Profit")
)
```

```python
from pyspark.sql.functions import sum, col

display(
    fact_vendor_performance
    .groupBy("VendorName")
    .agg(
        sum("GrossProfit").alias("Profit"),
        sum("TotalPurchaseQuantity").alias("Qty")
    )
    .withColumn(
        "ProfitPerUnit",
        col("Profit")/col("Qty")
    )
    .orderBy(col("ProfitPerUnit").desc())
)
```

```python
display(
    fact_vendor_performance
    .groupBy("VendorName")
    .avg("PurchasePrice","ActualPrice")
    .orderBy("avg(PurchasePrice)", ascending=False)
)
```

```python
from pyspark.sql.functions import col,sum

display(
    fact_procurement
    .groupBy("VendorName")
    .agg(
        sum("TotalFreight").alias("Freight"),
        sum("TotalPurchaseSpend").alias("Spend")
    )
    .withColumn(
        "FreightPct",
        col("Freight")/col("Spend")*100
    )
    .orderBy(col("FreightPct").desc())
)
```

```python
display(
    fact_vendor_performance
    .groupBy("VendorName")
    .sum("TotalSalesDollars")
    .orderBy("sum(TotalSalesDollars)")
)
```

```python
from pyspark.sql.functions import avg,sum

display(
    fact_vendor_performance
    .groupBy("VendorName")
    .agg(
        avg("StockTurnover").alias("Turnover"),
        sum("GrossProfit").alias("Profit")
    )
    .orderBy("Turnover", ascending=False)
)
```

```python
from pyspark.sql.functions import sum,avg

display(
    fact_inventory
    .groupBy("Description")
    .agg(
        sum("onHand").alias("Quantity"),
        avg("Price").alias("AvgPrice"),
        sum("InventoryValue").alias("InventoryValue")
    )
    .orderBy("InventoryValue", ascending=False)
)
```
